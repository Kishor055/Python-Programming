```python
#line Chart
import matplotlib.pyplot as plt

# Define the missing 'days' list (matching the 5 data points in a and b)
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

# Styling a line chart
a = [10, 20, 15, 25, 30]
b = [8, 14, 18, 20, 16]

plt.plot(
    days, a, color="blue", marker="o", linestyle="--", label="python"
)
plt.plot(days, b, color="green", marker="s", label="java")

plt.title("weekly score")
plt.xlabel("day")
plt.ylabel("score")  # Fixed from plot.ylabel to plt.ylabel
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Bars style 
subjects = ["Math", "Science", "English", "History","Marathi"]
marks = [85,78,92,70,98]
plt.bar(subjects, marks, color="steelblue")
plt.title("Marks by Subject")
plt.ylabel("Marks")
plt.show() 
```

```python
#ploting using pandas
import pandas as pd

df=pd.DataFrame({
    "name": ["Riya", "Aman","Neha",
             "Karan", "Isha", "Rohan"],
    "city" : ["Pune","Mumbai","Pune",
              "Kolhapur","Mumbai","Pune"],
    "marks" : [ 85, 62, 91, 45, 78, 70],
    "attendances" : [ 92, 75, 68, 84, 98, 89],
})
df.plot(x="name", y="marks",kind="bar",
        legend=False, title="Marks Of the Student")
plt.ylabel("Marks")
plt.show()
```

```python
df.groupby("city")["marks"].mean().plot(kind="bar", title="Average Marks by City")
plt.ylabel("Average Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Define your DataFrame
df = pd.DataFrame(
    {
        "name": ["Riya", "Aman", "Neha", "Karan", "Isha", "Rohan"],
        "city": ["Pune", "Mumbai", "Pune", "Kolhapur", "Mumbai", "Pune"],
        "marks": [85, 62, 91, 45, 78, 70],
        "attendances": [92, 75, 68, 84, 98, 89],
    }
)

# Plot using Seaborn
sns.set_theme(style="whitegrid")

sns.countplot(data=df, x="city")
plt.title("Student Per City")
plt.ylabel("Number of Students")
plt.xlabel("City")
plt.show()
```

```python
sns.scatterplot(
    data=df, 
    x="attendances", 
    y="marks", 
    hue="city", 
    s=120
)
plt.title("Attendance VS Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.show()
```

```python
sns.histplot(data=df, x="marks", bins=5, kde=True)
plt.title("Marks Distribution")
plt.show()
```

```python
sns.boxplot(data=df, x="city", y="marks")
plt.title("Marks By City")
plt.show()
```

```python
# Create column with matching name
df["game_hours"] = [4, 6, 3, 5, 2, 4]

# Select existing columns
corr = df[["marks", "attendances", "game_hours"]].corr()

sns.heatmap(
    corr, 
    annot=True, 
    fmt=".2f",
    cmap="coolwarm", 
    vmin=-1, 
    vmax=1
)
plt.title("Correlation")
plt.show()
```

```python
#plot the squares of the numbers 1 to 10 as a line chart with markers, title and labels.
import matplotlib.pyplot as plt

numbers = list(range(1, 11))
squares = [x**2 for x in numbers]

plt.plot(numbers, squares, marker="o", color="b", linestyle="-")

plt.title("Squares of Numbers (1 to 10)")
plt.xlabel("Number")
plt.ylabel("Square")

plt.grid(True)
plt.show()
```

```python
#Generates 500 Random numbers with np.random.normal and Draw a histogram with bins
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

data = np.random.normal(loc=0, scale=1, size=500)

plt.hist(data, bins=20, color="skyblue", edgecolor="black")

plt.title("Histogram of 500 Random Normal Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
#Create a figure with 2 rows and 2 columns of the subplots showing 4 different charts.
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y_line = [10, 20, 15, 25, 30]
categories = ["A", "B", "C", "D", "E"]
values = [5, 12, 19, 8, 15]
data_hist = np.random.normal(70, 10, 200)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, y_line, color="blue", marker="o", linestyle="--")
axes[0, 0].set_title("Line Chart")
axes[0, 0].set_xlabel("X-axis")
axes[0, 0].set_ylabel("Y-axis")

axes[0, 1].bar(categories, values, color="steelblue")
axes[0, 1].set_title("Bar Chart")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Values")


axes[1, 0].scatter(x, y_line, color="green", s=80)
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_xlabel("X-axis")
axes[1, 0].set_ylabel("Y-axis")


axes[1, 1].hist(data_hist, bins=10, color="teal", edgecolor="black")
axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# load any DataFrame and make a groupby bar chart using df.groupby(....).plot(kind="bar")
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame(
    {
        "city": ["Pune", "Mumbai", "Pune", "Kolhapur", "Mumbai", "Pune"],
        "marks": [85, 62, 91, 45, 78, 70],
    }
)

# Groupby and plot bar chart
df.groupby("city")["marks"].mean().plot(
    kind="bar", color="skyblue", edgecolor="black"
)

plt.title("Average Marks by City")
plt.xlabel("City")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```
